```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")



```

```python
# Creating a hypothetical Customer Churn Dataset

np.random.seed(42)

data = {
    'Customer_ID': range(1, 201),
    'Age': np.random.randint(18, 70, 200),
    'Monthly_Charges': np.random.randint(20, 120, 200),
    'Tenure_Months': np.random.randint(1, 72, 200),
    'Contract_Type': np.random.choice(['Monthly', 'One Year', 'Two Year'], 200),
    'Churn': np.random.choice(['Yes', 'No'], 200, p=[0.3, 0.7])
}

df = pd.DataFrame(data)

df.head()
```

```python
# Display basic information about the dataset

print("Dataset Shape:", df.shape)
print("\nDataset Information:")
df.info()
```

```python
# Statistical summary of numerical columns

df.describe()
```

```python
# Bar Chart: Churned vs Retained Customers

churn_counts = df['Churn'].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(churn_counts.index, churn_counts.values)

plt.title('Customer Churn Distribution')
plt.xlabel('Churn Status')
plt.ylabel('Number of Customers')

plt.show()
```

```python
# Churn Analysis by Contract Type

contract_churn = pd.crosstab(df['Contract_Type'], df['Churn'])

contract_churn.plot(kind='bar', figsize=(7, 5))

plt.title('Customer Churn by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)

plt.show()
```

```python
# Monthly Charges Distribution by Churn Status

plt.figure(figsize=(7, 5))

df.boxplot(column='Monthly_Charges', by='Churn')

plt.title('Monthly Charges by Churn Status')
plt.suptitle('')
plt.xlabel('Churn Status')
plt.ylabel('Monthly Charges')

plt.show()
```

```python
# Customer Age Distribution

plt.figure(figsize=(7, 5))

plt.hist(df['Age'], bins=10, edgecolor='black')

plt.title('Distribution of Customer Age')
plt.xlabel('Age')
plt.ylabel('Number of Customers')

plt.show()
```

```python
# Customer Tenure Distribution

plt.figure(figsize=(7, 5))

plt.hist(df['Tenure_Months'], bins=10, edgecolor='black')

plt.title('Distribution of Customer Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Number of Customers')

plt.show()
```

```python
# Calculate overall customer churn rate

churn_rate = (df['Churn'] == 'Yes').mean() * 100

print(f"Overall Customer Churn Rate: {churn_rate:.2f}%")
print(f"Customer Retention Rate: {100 - churn_rate:.2f}%")
```

```python
# Summary of key insights from the Customer Churn dataset

summary = pd.DataFrame({
    'Metric': [
        'Total Customers',
        'Churned Customers',
        'Retained Customers',
        'Overall Churn Rate (%)',
        'Average Monthly Charges',
        'Average Tenure (Months)'
    ],
    'Value': [
        len(df),
        (df['Churn'] == 'Yes').sum(),
        (df['Churn'] == 'No').sum(),
        round((df['Churn'] == 'Yes').mean() * 100, 2),
        round(df['Monthly_Charges'].mean(), 2),
        round(df['Tenure_Months'].mean(), 2)
    ]
})

summary
```

```python
# Final Project Conclusion

print("FINAL CUSTOMER CHURN ANALYSIS")
print("-" * 40)

print(f"Total Customers: {len(df)}")
print(f"Churned Customers: {(df['Churn'] == 'Yes').sum()}")
print(f"Retained Customers: {(df['Churn'] == 'No').sum()}")
print(f"Overall Churn Rate: {churn_rate:.2f}%")
print(f"Average Monthly Charges: {df['Monthly_Charges'].mean():.2f}")
print(f"Average Customer Tenure: {df['Tenure_Months'].mean():.2f} months")

print("\nKey Insight:")
print("Customer churn patterns can be analyzed using contract type,")
print("monthly charges, age, and customer tenure to support")
print("better customer retention strategies.")
```
